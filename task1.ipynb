```python
import pandas as pd

df = pd.read_excel("sales_noisy_500rows_10columns.xlsx")

print(df.head())
```

```python
print("Number of rows and columns:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values before cleaning:")
print(df.isnull().sum())
```

```python
print("Duplicate rows before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate rows after:", df.duplicated().sum())
```

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
number_columns = df.select_dtypes(include="number").columns

for col in number_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].fillna("Unknown")
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(df.duplicated().sum())

print("\nCleaned Dataset:")
print(df.head(10))
```

```python
df.to_excel("cleaned_sales_dataset.xlsx", index=False)
df.to_csv("cleaned_sales_dataset.csv", index=False)

print("Cleaned files saved successfully!")
```
